```python
df_data = pd.read_csv('seleksi bab 2.txt', index_col=0)
df_data.sort_values('C_6')
```

```python
df_C_only = df_data.drop(['Peneliti & Tahun'], axis=1)
df_C_only.head()
```

```python
df_calc = df_C_only.copy()
df_norm = pd.DataFrame()
kriteria_cols = list(config.keys())

for col, settings in config.items():
    if settings['tipe']=='benefit':
        df_norm[col] = df_calc[col]/df_calc[col].max()
    elif settings['tipe']=='cost':
        df_norm[col] = df_calc[col].min()/df_calc[col]

df_weighted = df_norm.copy()
for col, settings in config.items():
    df_weighted[col] = df_norm[col] * settings['bobot']
    
ideal_positif = {}
ideal_negatif = {}

for col in kriteria_cols:
    # Karena sudah dinormalisasi SAW (Benefit semua arahnya positif), 
    # maka Ideal Positif selalu Max, Negatif selalu Min
    ideal_positif[col] = df_weighted[col].max()
    ideal_negatif[col] = df_weighted[col].min()
    
def get_distance(row, ideal_dict):
        dist = 0
        for col in kriteria_cols:
            dist += (row[col] - ideal_dict[col]) ** 2
        return np.sqrt(dist)

df_calc['D_plus'] = df_weighted.apply(lambda x: get_distance(x, ideal_positif), axis=1)
df_calc['D_min'] = df_weighted.apply(lambda x: get_distance(x, ideal_negatif), axis=1)

df_calc['Skor_Akhir'] = df_calc['D_min'] / (df_calc['D_min'] + df_calc['D_plus'])

df_calc.sort_values(by='Skor_Akhir', ascending=False)
```
